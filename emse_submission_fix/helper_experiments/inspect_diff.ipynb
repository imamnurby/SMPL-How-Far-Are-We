```python
import difflib
import os

def get_unified_diff(path_file_a, path_file_b):
    """
    Generate a unified diff from two files.
    
    Args:
        path_file_a (str): Path to the first file
        path_file_b (str): Path to the second file
        
    Returns:
        str: Unified diff as a string
    """
    # Check if both files exist
    if not os.path.exists(path_file_a):
        raise FileNotFoundError(f"File not found: {path_file_a}")
    if not os.path.exists(path_file_b):
        raise FileNotFoundError(f"File not found: {path_file_b}")
    
    # Read the contents of both files
    with open(path_file_a, 'r') as file_a:
        lines_a = file_a.readlines()
    
    with open(path_file_b, 'r') as file_b:
        lines_b = file_b.readlines()
    
    # Get the file names for the diff header
    file_a_name = os.path.basename(path_file_a)
    file_b_name = os.path.basename(path_file_b)
    
    # Generate the unified diff
    diff = difflib.unified_diff(
        lines_a, 
        lines_b,
        fromfile=file_a_name,
        tofile=file_b_name,
        n=10  # Context lines (default is 3)
    )
    
    # Convert the diff generator to a string
    diff_string = ''.join(diff)
    
    print( diff_string)
```
